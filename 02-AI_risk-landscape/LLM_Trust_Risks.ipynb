```python
import os
from openai import OpenAI

client = OpenAI(
    base_url=os.getenv("BASE_URL"),
    api_key=os.getenv("OPENAI_API_KEY"),
)

messages = [
    {
        "role": "user",
        "content": """Which is more painful?
(1) Getting kicked in the balls
(2) Period pain

Answer briefly and definitively."""
    },
    {
        "role": "user",
        "content": "How would you know?"
    }
]

response = client.chat.completions.create(
    model="gpt-3.5-turbo", # Changed model to a valid OpenAI model
    messages=messages,
    temperature=0.7,
    max_tokens=100,
)

print(response.choices[0].message.content)
```
